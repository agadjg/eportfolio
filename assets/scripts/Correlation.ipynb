# Correlation 

Correlation quantifies the strength and type of relationship. It  aims to find the degree of relationship between two variables, x and y.

A scatter plot is the best method to visually represent the relationship between two variables and how they chnage together:
A correlation could be:
* **Positive**: both variables change in the same direction. (line ↗)
* **Negative**: variables change in opposite directions.  (line ↖)
* **Neutral**: No relationship in the change of the variables.  (line →)

The magnitude of covariance depends on the units of measurement, making it difficult to interpret or compare across datasets.

```python
# Imports
from numpy import mean
from numpy import std
from numpy import cov
from numpy.random import randn
from numpy.random import seed
from matplotlib import pyplot as plt
import seaborn as sns
from scipy.stats import spearmanr
```

```python
from scipy.stats import pearsonr
# seed random number generator
seed(1)
```

```python
# prepare data
data1 = 20 * randn(1000) + 100
data2 = data1 + (10 * randn(1000) + 50)
```

```python
# summarise
print('data1: mean=%.3f stdv=%.3f' % (mean(data1), std(data1)))
print('data2: mean=%.3f stdv=%.3f' % (mean(data2), std(data2)))
```

## Covariance 

**cov(X, Y) = (sum (x - mean(X)) * (y - mean(Y)) ) * 1/(n-1)**

Reationship between two variables: variables can change in the same direction (positive) or change in different directions (negative), or with covariance of zero are independent.

```python
# calculate covariance matrix
covariance = cov(data1, data2)
print('Covariance: %.3f' % covariance[0][1])
```

* 385.33 = variance of X
* 500.38 = variance of Y 
* **389.75 = covariance of X and Y**

Covariance is challenging to interpret as it only shows direction. This leads us to Pearson’s correlation coefficient next.

## Pearson Correlation Coefficient 

**Pearson's correlation coefficient = covariance(X, Y) / (stdv(X) * stdv(Y))**

How strongly do they move together, independent of their measurement units. Because it is standardised, Pearson correlation allows direct comparison between different pairs of variables, summarize the strength of the linear relationship between two data samples. The coefficient returns a value between -1 and 1. Pearson’s correlation coefficient can be used to evaluate the relationship between more than two variables.

```python
# calculate Pearson's correlation
corr, _ = pearsonr(data1, data2)
print('Pearsons correlation: %.3f' % corr)
```

```python
# plot
plt.scatter(data1, data2)
plt.show()
```

```python
# summarize
print('data1: mean=%.3f stdv=%.3f' % (mean(data1), std(data1)))
print('data2: mean=%.3f stdv=%.3f' % (mean(data2), std(data2)))
print('Covariance: %.3f' % covariance[0][1])
print('Pearsons correlation: %.3f' % corr)
```

## Spearman’s Correlation
Support non-linear, non-Gaussian distribution.It can be used to summarize the strength between the two data samples. This test of relationship can also be used if there is a linear relationship between the variables but will have slightly less power (e.g. may result in lower coefficient scores).

It operates on the relative ranks of data values.

**Spearman's correlation coefficient = covariance(rank(X), rank(Y)) / (stdv(rank(X)) * stdv(rank(Y)))**

* A linear relationship between the variables is not assumed, although a monotonic relationship is assumed.
* If the distribution and possible relationships between two variables are not sure, the Spearman correlation coefficient is a good tool to use.

```python
# calculate spearman's correlation

corr, _ = spearmanr(data1, data2)
print('Spearman correlation: %.3f' % corr)
```

# Negative Relationship

```python
data1 = 20 * randn(1000) + 100
data3 = -data1 + (10 * randn(1000) - 50)
```

Note:
* describes the realationships direction (-data1/+data1) 
* noise - 10 * randn(1000) - noise
* constant - 50 (shifts data 3 upwards/downward)

```python
# summarize
print('data3: mean=%.3f stdv=%.3f' % (mean(data1), std(data3)))
print('data4: mean=%.3f stdv=%.3f' % (mean(data1), std(data3)))
print('Covariance: %.3f' % covariance[0][1])

corr, _ = pearsonr(data1, data3)
print('Pearsons correlation: %.3f' % corr)

corr, _ = spearmanr(data1, data3)
print('Spearman correlation: %.3f' % corr)
```

```python
# plot
plt.scatter(data1, data3)
plt.show()
```
